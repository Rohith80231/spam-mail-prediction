Model Dependencies


```python
import numpy as np
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn. feature_extraction.text import TfidfVectorizer
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.metrics import accuracy_score
```

Data Preprocessing

```python
df = pd.read_csv('/content/mail_data.csv')
```

```python
print(df)
```

```python
#replace a null value with a null string
mail_data = df.where((pd.notnull(df)),'')
```

```python
#Printing first 5 rows
mail_data.head()
```

Label Encoding


```python
#label Encoding --> changing text to numerical
#ham-->1 and spam --> 0

mail_data.loc[mail_data['Category'] == 'spam', 'category',]=0
mail_data.loc[mail_data['Category'] == 'ham', 'category',]=1
```

```python
print(mail_data)
```

```python
X = mail_data['category']
Y = mail_data['Message']
```

```python
print(X)
```

```python
print(Y)
```

Splitting the data into training and testing data


```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=1)
```

Feature Extraction

```python
#Convertin test string to feature numerica vaues
feature_extraction = TfidfVectorizer(min_df = 1, stop_words='english', lowercase =True)

Y_train_features = feature_extraction.fit_transform(Y_train)
Y_test_features = feature_extraction.transform(Y_test)

X_train = X_train.astype('int')
X_test = X_test.astype('int')
```

```python
print(Y_train_features)
```

```python
model = LogisticRegression()
```

```python
model.fit(Y_train_features,X_train)
```

```python
prediction_on_training_data = model.predict(Y_train_features)
accuracy_on_training_data = accuracy_score(X_train,prediction_on_training_data)
print(accuracy_on_training_data)
```

```python
prediction_on_testing_data = model.predict(Y_test_features)
accuracy_on_testing_data = accuracy_score(X_test,prediction_on_testing_data)
print(accuracy_on_testing_data)
```

```python
input_mail = ["WINNER!! As a valued network customer you have been selected to receivea £900 prize reward! To claim call 09061701461. Claim code KL341. Valid 12 hours only."]

input_data_features = feature_extraction.transform(input_mail)
prediction = model.predict(input_data_features)

print(prediction)
```
